### 🔧 Robot (จากสัปดาห์ที่แล้ว — รันก่อนเริ่ม)

````python
# 🛟 Recovery Cell
class Robot:
    def __init__(self, name, battery=100):
        self.name = name
        self.battery = battery

    def __str__(self):
        return f"🤖 {self.name} [{self.battery}%]"

    def move(self, distance):
        cost = distance // 10
        if cost > self.battery:
            return False
        self.battery -= cost
        return True

    def charge(self, amount):
        self.battery = min(self.battery + amount, 100)
        return self.battery

    def work(self):
        return f"{self.name} ทำงานทั่วไป"
````
