### 📚 เรียนก่อนทำ Level 2: closure + `nonlocal` (8 นาที)

function สร้าง function ข้างในแล้ว return ออกมาได้ — function ข้างใน **จำค่าตัวแปรของชั้นนอก** ไว้ได้ แม้ชั้นนอกทำงานจบแล้ว เรียกว่า **closure**

````python
def make_multiplier(n):
    def multiply(x):
        return x * n          # จำ n ไว้
    return multiply           # คืนตัว function (ไม่มีวงเล็บ)

double = make_multiplier(2)
triple = make_multiplier(3)
print(double(10), triple(10))
````

**ถ้าอยาก "แก้" ค่าของชั้นนอก** ต้องประกาศ `nonlocal` ไม่งั้น Python คิดว่าเป็นตัวแปรใหม่ในชั้นใน → `UnboundLocalError`

🤔 **ทายก่อนรัน:** `c1(), c1(), c1(), c2()` ได้อะไร?

````python
def make_counter():
    count = 0
    def click():
        nonlocal count
        count += 1
        return count
    return click

c1 = make_counter()
c2 = make_counter()
print(c1(), c1(), c1(), c2())     # c1 กับ c2 มี count คนละตัว
````

### 📚 เรียนก่อนทำ Level 3: decorator แรก (8 นาที)

**decorator = function ที่รับ function เข้าไป แล้วคืน function ใหม่ (wrapper) ที่ทำอะไรเพิ่มก่อน/หลัง** โดยไม่แก้โค้ดของ function เดิม
🎁 เหมือนห่อของขวัญ: ของข้างในเหมือนเดิม แต่มีกล่องห่อ

นี่คือการบ้าน `@shout` เมื่อสัปดาห์ก่อน — ดูอีกครั้ง:

````python
def shout(func):
    def wrapper():
        return func().upper() + "!"     # เรียกของเดิม แล้วแต่งผล
    return wrapper

def hello():
    return "hello robot"

hello = shout(hello)        # ห่อด้วยมือ
print(hello())
````

`@shout` ที่วางบน `def` คือการเขียนย่อของ `hello = shout(hello)` **เหมือนกันทุกประการ**

````python
@shout
def bye():
    return "bye robot"

print(bye())
````

**3 กฎของ wrapper**
1. ต้อง **เรียก** `func()` (มีวงเล็บ)
2. ต้อง **return** ผลลัพธ์ ไม่งั้นได้ `None`
3. decorator ต้อง **return wrapper** (ไม่มีวงเล็บ)

### 📚 เรียนก่อนทำ Level 4: decorator ที่ใช้ได้กับทุก function (8 นาที)

wrapper ของ Level 3 รับ argument ไม่ได้เลย ถ้าเอาไปห่อ `move(direction, steps)` จะพัง
**`*args`** = รับ argument ตามตำแหน่งกี่ตัวก็ได้ (เป็น tuple) · **`**kwargs`** = รับแบบ `ชื่อ=ค่า` (เป็น dict)

````python
def show(*args, **kwargs):
    print("args =", args)
    print("kwargs =", kwargs)

show(1, "x", name="Bolt", speed=5)
````

ใน wrapper: รับด้วย `*args, **kwargs` แล้ว **ส่งต่อ** ให้ `func(*args, **kwargs)` ครบทุกตัว

**`@wraps(func)`** — ไม่ใส่ ชื่อ function จะกลายเป็น `wrapper` ▶️ ดูความต่าง

````python
from functools import wraps

def plain(func):
    def wrapper(*args, **kwargs):
        return func(*args, **kwargs)
    return wrapper

def nice(func):
    @wraps(func)
    def wrapper(*args, **kwargs):
        return func(*args, **kwargs)
    return wrapper

@plain
def a(): pass

@nice
def b(): pass

print(a.__name__, b.__name__)
````

# 🔹 Level 5

### 🛟 Recovery Cell — `blackbox` + `wraps` (รันถ้า Level 4 ยังไม่เสร็จ)

````python
# 🛟 Recovery Cell
from functools import wraps

def blackbox(func):
    @wraps(func)
    def wrapper(*args, **kwargs):
        parts = [repr(a) for a in args] + [f"{k}={v!r}" for k, v in kwargs.items()]
        print(f"📼 {func.__name__}({', '.join(parts)})")
        result = func(*args, **kwargs)
        print(f"📼 → {result!r}")
        return result
    return wrapper
````

### 📚 เรียนก่อนทำ Level 5: decorator ที่รับค่า — 3 ชั้น (10 นาที)

`@repeat(3)` มีวงเล็บ → Python **เรียก `repeat(3)` ก่อน** ได้ decorator ออกมา แล้วค่อยเอาไปห่อ function

```text
repeat(times)            ← ชั้น 1: รับค่าตั้งค่า
  └─ decorator(func)     ← ชั้น 2: รับ function (เหมือน Level 3–4)
       └─ wrapper(...)   ← ชั้น 3: ทำงานจริงตอนเรียก
```

````python
def repeat(times):
    def decorator(func):
        @wraps(func)
        def wrapper(*args, **kwargs):
            result = None
            for _ in range(times):
                result = func(*args, **kwargs)
            return result
        return wrapper
    return decorator

@repeat(3)                 # = beep = repeat(3)(beep)
def beep():
    print("บี๊บ!")

beep()
````

**wrapper ตัดสินใจได้ว่าจะเรียก function จริงหรือไม่** — ถ้าไม่ผ่านเงื่อนไข return ข้อความอื่นไปเลย (Level 5 ใช้ท่านี้)

```python
        ...                   # ล้มเหลว → วนรอบต่อไป
return "❌ ยอมแพ้"            # ครบทุกรอบแล้ว (นอก loop)
```
> ไม่มีโจทย์จิ๋ว — โครงด้านบนคือ 80% ของ Level 6 แล้ว ใส่ใน decorator 3 ชั้นได้เลย

## 🎯 Level 6: ระบบลองใหม่ 🔁

**หัวข้อที่ฝึก:** decorator + exception

### ภารกิจ
สัญญาณวิทยุไม่เสถียร บางครั้งส่งไม่สำเร็จ (เกิด `ConnectionError`)

### ข้อกำหนด
- `retry(times)` → ถ้า function เกิด `ConnectionError` ให้แสดง `⚠️ ครั้งที่ n ล้มเหลว: <ข้อความ>` แล้วลองใหม่ สูงสุด `times` ครั้ง
- ครบแล้วยังไม่สำเร็จ → return `"❌ ยอมแพ้"`

# 🔹 Level 7

### 📚 Level 7: ไม่มีเรื่องใหม่ — เช็กลิสต์ 5 บั๊กยอดฮิตของ decorator
| # | ตรวจ | ถูก |
|:-:|---|---|
| 1 | wrapper รับ argument ครบไหม | `wrapper(*args, **kwargs)` |
| 2 | **เรียก** function เดิมหรือแค่อ้างถึง | `func(*args, **kwargs)` |
| 3 | **return** ผลลัพธ์หรือยัง | `return result * 2` |
| 4 | ชื่อ function หายไหม | `@wraps(func)` |
| 5 | decorator คืนตัว wrapper หรือเผลอเรียก | `return wrapper` (ไม่มีวงเล็บ) |

> 🔍 แก้ทีละจุด รันทีละครั้ง ดูว่า error เปลี่ยนไปยังไง

## 🎯 Level 7: 🐛 อุปกรณ์ช็อต!

**หัวข้อที่ฝึก:** debug decorator

### ภารกิจ
decorator `double_result` ควรคูณผลลัพธ์ด้วย 2 แต่มี **บั๊ก 5 จุด**

### Expected Output
```text

```

### 💡 คำใบ้ (เปิดทีละขั้น ลองคิดเองก่อนนะ!)

<details><summary>ใบ้ 1: เริ่มยังไง</summary>

รันก่อน ดู error แรก · เริ่มที่บรรทัด `return wrapper()`

</details>

<details><summary>ใบ้ 2: ใช้เครื่องมืออะไร</summary>

ใช้เช็กลิสต์ 5 ข้อด้านบนไล่ทีละบรรทัด

</details>

<details><summary>ใบ้ 3: เกือบเฉลย</summary>

`def wrapper(*args, **kwargs):` · `result = func(*args, **kwargs)` · `return result * 2` · `@wraps(func)` · `return wrapper`

</details>


📖 ดูเนื้อหาที่เกี่ยวข้องได้ใน `Student_guide.md`

````python
def double_result(func):
    def wrapper(x):
        result = func
        result * 2
    return wrapper()

@double_result
def power(base, exp=2):
    return base ** exp

print(power(3))
print(power(2, exp=5))
print(power.__name__)
````

# 🔹 Level 8

### 📚 เรียนก่อนทำ Level 8: ซ้อน decorator + ใช้กับ method (10 นาที)

**ซ้อนหลายชั้น:** ตัว **บนสุดห่อนอกสุด** → ทำงานก่อน
```python
@a
@b
def f(): ...
# เท่ากับ f = a(b(f))
```
🤔 **ทายก่อนรัน:** บรรทัดไหนออกมาก่อน?

````python
def tag(name):
    def decorator(func):
        @wraps(func)
        def wrapper(*args, **kwargs):
            print(f"เข้า {name}")
            return func(*args, **kwargs)
        return wrapper
    return decorator

@tag("A")
@tag("B")
def hello():
    print("ตัวจริง")

hello()
````

**decorator บน method:** argument ตัวแรกที่ wrapper ได้รับคือ **`self`** — จึงอ่าน `self.battery` / `self.name` ได้ใน wrapper

````python
def show_self(func):
    @wraps(func)
    def wrapper(self, *args, **kwargs):
        print(f"[{self.name}] เรียก {func.__name__}")
        return func(self, *args, **kwargs)
    return wrapper

class Bot:
    def __init__(self, name):
        self.name = name

    @show_self
    def wave(self):
        return "👋"

print(Bot("Zip").wave())
````

**`getattr(obj, "ชื่อ")`** ดึง method ด้วยข้อความ → `getattr(bolt, "scan")()` = `bolt.scan()`
